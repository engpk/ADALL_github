```python
import pandas as pd
# Example: Replace this with the raw URL of your GitHub file
github_raw_url = 'https://raw.githubusercontent.com/engpk/ADALL_github/refs/heads/main/laptop_prices_2024_sgd_TL.csv'
try:
    df = pd.read_csv(github_raw_url)
    print("Successfully loaded data from GitHub!")
    display(df.head())

    #import the plt library
    import matplotlib.pyplot as plt

    #for each numerical column, display a histogram of that column
    for col in df.select_dtypes(include=['number']).columns:
        df[col].hist()
        plt.title(col)
        plt.show()

except Exception as e:
    print(f"Error loading data: {e}")
    print("Please ensure the URL is correct and the file format is compatible with `pd.read_csv`.")
```
